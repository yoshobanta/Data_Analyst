```python
arr = np.array([[1,2,3],[4,5,6]])

print("Array: \n",arr)

print("Shape :" , arr.shape)

print("Number of dimentions : " , arr.ndim)

print("Size(number of elements) :" ,arr.size)

print("Data Type :", arr.dtype)  #(o/p may vary based on the platform)

print("Item size (in bytes) : ",arr.itemsize)  #(o/p may vary based on the platform)
```

```python
# Numpy Vectorize Operation 

arr1 =np.array([1,2,3,4,5])
arr2 =np.array([10,20,30,40,50])

# Elements wise addition
print("Addition :" , arr1 + arr2)

# Elemnts wise substraction
print("Substraction :" , arr2 - arr1)
print("mul :" , arr1 * arr2)
print("div :" , arr2 / arr1)
```

```python
# Universal Function 
arr = np.array([1,2,3,4,5])

#square root
print(np.sqrt(arr))

#exponential
print(np.exp(arr))     #exponential function e^^x


# e ≈ 2.71828 (Euler’s number, the base of natural logarithms).

# For each element in arr, NumPy calculates e^^value

#sine
print(np.sin(arr))

#natural log   means base is e
print(np.log(arr))
```

```python
# Stastical Concepts - Normalization
# have a mean of 0 and standard deviation of 1

data = np.array([1,2,3,4,5])

# Calculate the mean and standard deviation

mean = np.mean(data)
std_dev = np.std(data)
print(mean) 
print(std_dev) 

# Normalized the data
normalize_data = (data - mean) / std_dev     # standard normal distribution or standardiation 

print("Normalized data : " , normalize_data)
```

```python
data = np.arange(1,10)

mean = np.mean(data)
median = np.median(data)
std_dev = np.std(data)
variance = np.var(data)

print(mean,median,std_dev,variance , sep = " :: ") 
```

```python
# Logical Operations
data = np.arange(1,11)

print(data)
print(data > 5)
print(data[data > 5])


# print(data[data >= 5  and  data <= 8])

# here and will not work . In numpy & will work and it is overloaded to work with each elemnet of array.
# and will not work as it does not know how to deal with 2d array and all .

print(data[(data >= 5)  & (data <= 8)])
```
