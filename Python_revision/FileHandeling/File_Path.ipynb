```python
import os

print(os.getcwd())
```

```python
# Create new directiory

new_dir = 'package'
os.mkdir(new_dir)

print(f'Directory {new_dir} is been created')
```

```python
# Listing Files and Directories

# syntax = os.listdir(path)    '.' menns current '..' means parent 

items = os.listdir('.')     #' . ' it is used for current working directory.
print(items)
```

```python
# Joining Paths

dir_name = 'folder'
file_name = 'file.txt'
full_path = os.path.join(dir_name,file_name)
print(full_path)
```

```python
dir_name = 'folder'
file_name = 'file.txt'
full_path = os.path.join(os.getcwd(),dir_name,file_name)
print(full_path)
```

```python
path = 'example1.txt'

if os.path.exists(path):
    print(f"The path '{path}' exist")
else:
    print(f"The path '{path}' does not exist")
```

```python
# Checking if the file path is a file or a directory

file_path = 'example.txt'

if os.path.isfile(file_path):
    print(f"The path '{file_path}' is a file")
elif os.path.isdir(file_path):
    print(f"The path '{file_path}' is a directory")
else :
    print("Neither a file nor a directory")
    
```

```python
# Getting absolute path

relative_path = 'example.txt'

absolute_path = os.path.abspath(relative_path)
print(absolute_path)


```
