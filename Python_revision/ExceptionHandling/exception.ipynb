Exception Handling in python allows us to handles errors gracefully and take corrective action  without stoping the execution of the program.

Exception are the events that disrupt the normal flow of the program . They occur when an error is encountred during program execution . Common exception includes :-

ZeroDivisionError
FileNotFoundError
ValueError
TypeError

```python
# try ,except , else

try:
    num = int(input("enter a number"))
    result = 10/num
    
except ValueError:
    print("Enter a proper number")
    
except ZeroDivisionError as ex:
    print(ex)
    print('Please enter denominator')
    
except Exception as ex:
    print(ex)
    
else:
    print(f"The result is {result}")
```

```python
# try ,except , else and finally

try:
    num = int(input("enter a number"))
    result = 10/num
    
except ValueError:
    print("Enter a proper number")
    
except ZeroDivisionError as ex:
    print(str(ex) + ' error')
    print('Please enter denominator as non zero to correct it')
    
except Exception as ex:
    print(ex)
    
else:
    print(f"The result is {result}")
    
finally: 
    print('Execution completed')


# suppose u connect to a database and do some operations then u want to close the connection 
# in this scenirio finally will be used .
```
