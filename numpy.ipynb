# 📘 NumPy Notes

## 🔹 Introduction
**NumPy (Numerical Python)** is a powerful library for numerical computing in Python. It provides support for:
- Multidimensional arrays
- Mathematical operations
- Linear algebra
- Random number generation

---

## 🔹 Importing NumPy

```python
import numpy as np
````

---

## 🔹 Creating Arrays

### 1. From Lists or Tuples

```python
np.array([1, 2, 3])
np.array([[1, 2], [3, 4]])
```

### 2. Predefined Arrays

```python
np.zeros((2, 3))        # 2x3 array of zeros
np.ones((2, 3))         # 2x3 array of ones
np.eye(3)               # 3x3 Identity matrix
np.full((2, 2), 5)      # 2x2 array filled with 5
np.arange(0, 10, 2)     # [0, 2, 4, 6, 8]
np.linspace(0, 1, 5)    # 5 values between 0 and 1
```

---

## 🔹 Array Attributes

```python
arr.shape      # Dimensions (rows, cols)
arr.ndim       # Number of dimensions
arr.size       # Total number of elements
arr.dtype      # Data type of elements
arr.itemsize   # Size of one element (in bytes)
```

---

## 🔹 Indexing and Slicing

```python
arr[1]           # 2nd element
arr[1:4]         # Elements from index 1 to 3
arr[:, 1]        # All rows, 2nd column
arr[1, :]        # 2nd row, all columns
arr[-1]          # Last row
```

---

## 🔹 Array Operations

### Arithmetic

```python
a + b
a - b
a * b
a / b
a ** 2
```

### Matrix Operations

```python
a.T                      # Transpose
np.dot(a, b)             # Matrix multiplication
np.linalg.inv(a)         # Inverse of matrix
np.linalg.det(a)         # Determinant
```

---

## 🔹 Aggregate Functions

```python
np.sum(arr)
np.min(arr)
np.max(arr)
np.mean(arr)
np.std(arr)
np.var(arr)
np.median(arr)
np.cumsum(arr)   # Cumulative sum
np.cumprod(arr)  # Cumulative product
```

---

## 🔹 Reshaping and Flattening

```python
arr.reshape(2, 3)
arr.flatten()
arr.ravel()
```

---

## 🔹 Stacking Arrays

```python
np.hstack((a, b))         # Horizontal stack
np.vstack((a, b))         # Vertical stack
np.concatenate((a, b), axis=0)  # Along rows
```

---

## 🔹 Useful Functions

```python
np.unique(arr)            # Unique values
np.sort(arr)              # Sort array
np.where(arr > 5)         # Indices where condition met
np.any(arr > 0)           # Any True?
np.all(arr > 0)           # All True?
```

---

## 🔹 Random Module

```python
np.random.rand(2, 2)         # Uniform distribution [0, 1)
np.random.randn(2, 2)        # Standard normal distribution
np.random.randint(0, 10, 5)  # 5 random ints [0,10)
np.random.seed(42)           # Seed for reproducibility
```

---

## 🔹 Broadcasting

```python
a = np.array([1, 2, 3])
b = 2
a * b   # [2, 4, 6]
```
